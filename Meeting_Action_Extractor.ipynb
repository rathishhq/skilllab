```python

# Install and import required libraries
%pip install -q ipywidgets

import ipywidgets as widgets
from IPython.display import display

# Sample notes (used if not already defined)
if "sample_notes" not in globals():
    sample_notes = """
Meeting Notes:

Chetan will complete the project report by Friday.
Arun will prepare the presentation by Monday.
The team will review the budget next Wednesday.
Priya will send the meeting summary by Thursday.
"""

# -----------------------------------------
# 1. NOTES INPUT
# -----------------------------------------

notes_box = widgets.Textarea(
    value=sample_notes,
    description="Notes:",
    layout=widgets.Layout(width="95%", height="170px"),
    style={"description_width": "initial"}
)

# -----------------------------------------
# 2. MODE SELECTION
# -----------------------------------------

mode_box = widgets.ToggleButtons(
    options=[
        ("Offline", "offline"),
        ("GEMINI FLASH 3.0", "live")
    ],
    value="offline",
    description="Mode:"
)

# -----------------------------------------
# 3. EXTRACT BUTTON
# -----------------------------------------

button = widgets.Button(
    description="Extract Action Items",
    button_style="primary",
    icon="check",
    layout=widgets.Layout(width="220px")
)

output = widgets.Output()

# -----------------------------------------
# 4. BUTTON FUNCTION
# -----------------------------------------


def on_extract_clicked(_):
    with output:
        output.clear_output(wait=True)

        text = notes_box.value.strip()

        if not text:
            print("Please enter meeting notes first.")
            return

        selected_mode = mode_box.value

        print(
            "Mode:",
            "OFFLINE" if selected_mode == "offline"
            else "LIVE GEMINI API"
        )

        try:
            if selected_mode == "offline":
                actions = offline_extract(text)
            else:
                actions = live_extract(text)

        except Exception as exc:
            import traceback
            print(f"ERROR: {type(exc).__name__}: {exc}")
            traceback.print_exc()
            return

        if not actions:
            print("No action items found.")
            return

        print("\nEXTRACTED ACTION ITEMS")
        print("=" * 40)

        for i, action in enumerate(actions, 1):
            print(f"\n{i}. {action['task']}")
            print(f"   Owner: {action['owner']}")
            print(f"   Due: {action['deadline']}")

        print("\nPlease review the extracted items.")

display(
    widgets.VBox([
        notes_box,
        mode_box,
        button,
        output
    ])
)
```

```python

import re

def offline_extract(text):
    actions = []

    for line in text.splitlines():
        line = line.strip()

        if not line:
            continue

        # Look for lines that sound like action items
        if re.search(
            r"\b(need to|will|must|should|assign|prepare|complete|send|submit|review|finish|follow up)\b",
            line,
            re.IGNORECASE
        ):
            actions.append({
                "task": line,
                "owner": "Not specified",
                "deadline": "Not specified"
            })

    return actions
```

```python
button.on_click(on_extract_clicked)
display(button, output)
```
